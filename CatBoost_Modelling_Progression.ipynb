Employment and Income Modeling with CatBoost


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import scipy.stats as stats
import warnings
from catboost import CatBoostClassifier, Pool
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, f1_score
```

```python
nsduh_2021_2024 = pd.read_csv('C:/Users/jkirk/Downloads/nsduh_2021_2024.csv')
```

```python
X = nsduh_2021_2024[["WRKSTATWK2", "IRFAMIN3", "GOVTPROG"]]
y = nsduh_2021_2024["MRJFLAG"]

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75)

model = CatBoostClassifier(iterations=200, depth=4, silent=True)
model.fit(X_train, y_train, cat_features=["WRKSTATWK2", "IRFAMIN3", "GOVTPROG"])

print("Test accuracy:", model.score(X_test, y_test))
print(model.get_feature_importance)

y_pred = model.predict(X_test)

f1 = f1_score(y_test, y_pred)
print(f"Binary F1 Score: {f1:.4f}")
```
